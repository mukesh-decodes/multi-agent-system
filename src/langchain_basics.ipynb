```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from utils.structured_response import Answer
from utils.promt_template import prompt_template_for_country
from time import perf_counter
load_dotenv('.env')

start=perf_counter()
llm=ChatOpenAI(model="gpt-5-nano",temperature=0)
# response=llm.invoke("Explain me  AI in 5 lines max.")

# structure_llm=llm.with_structured_output(Answer)

# response =structure_llm.invoke(" Explain what RAG is and why it is useful in 5 lines.")

# print(response)
# print(type(response))

messages =prompt_template_for_country.invoke({
    "country":"Russia",
    "state":"poor"
})

structured_llm=llm.with_structured_output(Answer)
response=structured_llm.invoke(messages)

end=perf_counter()
print(f"time to response {end-start:.4f} sec")
print("Response--->\n",response)
print("Type ==>\n",type(response))

```

```python
start=perf_counter()
chain_runnable=prompt_template_for_country|structured_llm # LangChain Expression Language (LCEL)

response_from_chain=chain_runnable.invoke({
    "country":"India",
    "state":"culturely rich compared to all the countries of the world"

})
end=perf_counter() # from time
print("response_from_chain=>" ,response_from_chain)
print(f"timer 2 {end-start:.4f}")
```
