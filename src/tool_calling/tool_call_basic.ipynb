```python


from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
from pydantic import BaseModel ,Field

class AnswerWithTools(BaseModel):
    ans:str=Field(description="String of tokens from model in repsonse")
    topic: list[str] =Field(description="main topics covered")
    tools: list[str] = Field(description="Names of tools used")

```

```python
@tool(description="multiplies two numbers")
def guna(a:int,b:int)->int:
    """Multipy two number"""
    return a*b
```

```python
llm = ChatOpenAI(model="gpt-5-nano", temperature=0)
llm_with_tools = llm.bind_tools([guna], tool_choice="guna")


```

```python
# Ask the model to call the tool, then execute the requested call.

res = llm_with_tools.invoke("What is the product of 21 and 35?")

print(res.tool_calls)

for tool_call in res.tool_calls:
    if tool_call["name"] == guna.name:
        result = guna.invoke(tool_call["args"])
        print(f"{tool_call['name']} returned {result}")
```
